## Normalisasi

Dalam citra Nilai intensitas itu berada di 0 - 255

pada umumnya orang atau pada metode image processin dia melakukan conversi nilai menjadi 0 - 1. Hal ini disebnut dengan normalisasi

0 -> 0

255 -> 1

Dalam pengolahan citra, normalisasi adalah proses untuk mengganti jangkauan nilai intensitas piksel citra. Kegunaannya antara lain perbaikan foto berkontras rendah akibat terkena sinar berlebih. Normalisasi terkadang disebut dengan perentangan kontras atau perentangan histogram.

Normalisasi adalah pemetaan citra berderajat keabuan berdimensi n dengan rentang nilai [min, maks] ke citra dengan rentang nilai [minBaru, maksBaru].

## Min-Max Normalization (0–1)

```python
import numpy as np

img = np.array([[0, 128, 255],[22, 50, 60],[90, 150, 200]], dtype=np.float32)
normalized = (img - img.min()) / (img.max() - img.min())
print(img)
print(normalized)
```

```python
import matplotlib.pyplot as plt
plt.imshow(img, 'gray')
```

```python
plt.imshow(normalized, 'gray')
```

# 2. Standardization (Z-score)

Z score adalah ukuran statistik yang menunjukkan penyimpangan suatu titik data dari rata-rata dalam satuan standar deviasinya.

Standar deviasi (atau simpangan baku) adalah ukuran statistik yang menunjukkan seberapa menyebar atau bervariasinya sekumpulan data terhadap nilai rata-ratanya (mean). 

```python
mean= img.mean()
std = img.std()
print(mean)
print(std)
standardized = (img - mean) / std
print(standardized)
```

```python
import numpy as np
from scipy.signal import convolve2d
import matplotlib.pyplot as plt
from skimage import io, color

# Baca gambar & ubah ke grayscale
img = color.rgb2gray(io.imread("OIP.jpeg"))

# Kernel (contoh: edge detection sederhana)
kernel = np.array([[-1, -1, -1],
                   [-1,  8, -1],
                   [-1, -1, -1]])

# Convolution
convolved = convolve2d(img, kernel, mode='same', boundary='symm')

# Tampilkan hasil
fig, axes = plt.subplots(1, 2, figsize=(10, 5))
axes[0].imshow(img, cmap='gray')
axes[0].set_title("Original Grayscale")
axes[0].axis("off")

axes[1].imshow(convolved, cmap='gray')
axes[1].set_title("After Convolution (Edge Detection)")
axes[1].axis("off")

plt.show()
```
