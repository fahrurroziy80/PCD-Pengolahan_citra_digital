```python
import numpy as np
from skimage import color
import math
from skimage import filters, util
import matplotlib.pyplot as plt
from skimage import io, img_as_float
from skimage.restoration import denoise_bilateral, denoise_wavelet
from skimage.filters import median, gaussian
from skimage.morphology import disk
from skimage.metrics import mean_squared_error, peak_signal_noise_ratio

print("Library import sukses ✅")
```

# Load Gambar Asli dan Gambar Noisy

```python
# Membaca gambar asli dan noisy
img_gt = io.imread('01.jpg')
img_noise = io.imread('noise01.jpg')

# Tampilkan kedua gambar
fig, ax = plt.subplots(1, 2, figsize=(12, 6))
ax[0].imshow(img_gt)
ax[0].set_title("Gambar Asli")
ax[1].imshow(img_noise)
ax[1].set_title("Gambar Noisy")
plt.show()
```

```python
gauss_filtered = gaussian(img_noise, sigma=1)
plt.imshow(gauss_filtered, cmap='gray')
plt.title('Gaussian Filter')
plt.show()
```

```python
plt.subplots(figsize=(10,10))
plt.imshow(img_noise,'gray');
```

```python
img_filtered = gaussian(img_noise, sigma=2)
```

```python
plt.subplots(figsize=(10,10))
plt.imshow(img_filtered,'gray');
```
