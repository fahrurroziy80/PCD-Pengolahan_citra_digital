# 🧠 Penghilangan Noise pada Citra Menggunakan Gaussian, Median, dan Bilateral Filter

```python

import cv2
import numpy as np
import matplotlib.pyplot as plt
from skimage import io, color, util, filters, restoration
from skimage.metrics import peak_signal_noise_ratio, mean_squared_error
```

```python

# Membaca gambar asli dan gambar dengan noise
img_gt = io.imread('01.jpg')
img_noise = io.imread('noise01.jpg')

# Konversi ke grayscale
img_gt_gray = color.rgb2gray(img_gt)
img_noise_gray = color.rgb2gray(img_noise)

# Tampilkan kedua gambar
fig, ax = plt.subplots(1, 2, figsize=(10, 5))
ax[0].imshow(img_gt_gray, cmap='gray')
ax[0].set_title('Gambar Asli')
ax[0].axis('off')

ax[1].imshow(img_noise_gray, cmap='gray')
ax[1].set_title('Gambar dengan Noise')
ax[1].axis('off')

plt.show()
```

```python

# Gaussian Filter
gaussian_filtered = filters.gaussian(img_noise_gray, sigma=1)

plt.imshow(gaussian_filtered, cmap='gray')
plt.title('Gaussian Filter Result')
plt.axis('off')
plt.show()
```

```python

# Median Filter
img_noise_ubyte = util.img_as_ubyte(img_noise_gray)
median_filtered = filters.median(img_noise_ubyte)

plt.imshow(median_filtered, cmap='gray')
plt.title('Median Filter Result')
plt.axis('off')
plt.show()
```

```python

# Bilateral Filter
bilateral_filtered = restoration.denoise_bilateral(
    img_noise_gray, sigma_color=0.05, sigma_spatial=15, channel_axis=None)

plt.imshow(bilateral_filtered, cmap='gray')
plt.title('Bilateral Filter Result')
plt.axis('off')
plt.show()
```

```python

# Pastikan semuanya dalam range [0,1]
img_gt_gray = util.img_as_float(img_gt_gray)
gaussian_filtered = util.img_as_float(gaussian_filtered)
median_filtered = util.img_as_float(median_filtered)
bilateral_filtered = util.img_as_float(bilateral_filtered)

# Hitung MSE dan PSNR
mse_gaussian = mean_squared_error(img_gt_gray, gaussian_filtered)
psnr_gaussian = peak_signal_noise_ratio(img_gt_gray, gaussian_filtered)

mse_median = mean_squared_error(img_gt_gray, median_filtered)
psnr_median = peak_signal_noise_ratio(img_gt_gray, median_filtered)

mse_bilateral = mean_squared_error(img_gt_gray, bilateral_filtered)
psnr_bilateral = peak_signal_noise_ratio(img_gt_gray, bilateral_filtered)

# Tampilkan hasil evaluasi
print("📊 Evaluasi MSE dan PSNR:")
print(f"Gaussian Filter  -> MSE: {mse_gaussian:.4f}, PSNR: {psnr_gaussian:.2f} dB")
print(f"Median Filter    -> MSE: {mse_median:.4f}, PSNR: {psnr_median:.2f} dB")
print(f"Bilateral Filter -> MSE: {mse_bilateral:.4f}, PSNR: {psnr_bilateral:.2f} dB")
```
