# Deskripsi Kode: Membaca dan Menampilkan Gambar dengan Python

Kode berikut digunakan untuk membaca file gambar (`.JPG`) dan menampilkannya menggunakan **scikit-image** dan **matplotlib**.
Kode tersebut digunakan untuk membaca dan menampilkan gambar dengan Python.

import skimage dan import skimage.io → mengimpor library scikit-image untuk memproses gambar.

import matplotlib.pyplot as plt → mengimpor library Matplotlib untuk menampilkan gambar.

print("import berhasil") → memastikan bahwa library berhasil diimpor.

image = skimage.io.imread("IMG_3570.JPG") → membaca file gambar IMG_3570.JPG dan menyimpannya dalam variabel image.

plt.imshow(image) → menampilkan gambar tersebut dalam jendela plot Matplotlib.

Singkatnya, kodingan ini fungsinya membaca file gambar dan menampilkannya ke layar.

```python
import skimage
import skimage.io
import matplotlib.pyplot as plt

print("import berhasil")

image = skimage.io.imread("IMG_3570.JPG")

plt.imshow(image)
```

from skimage.color import rgb2gray
→ Mengimpor fungsi rgb2gray dari modul skimage.color.
→ Fungsi ini digunakan untuk mengubah gambar berwarna (RGB) menjadi citra grayscale (hitam putih).

grayscale = rgb2gray(image)
→ Mengonversi variabel image (yang berisi gambar hasil pembacaan sebelumnya) menjadi grayscale.
→ Hasil konversi disimpan dalam variabel grayscale.

plt.imshow(grayscale, 'gray')
→ Menampilkan gambar grayscale menggunakan Matplotlib.
→ Argumen 'gray' dipakai agar warna ditampilkan dalam mode hitam putih.

```python
from skimage.color import rgb2gray

grayscale = rgb2gray(image)
plt.imshow(grayscale,'gray')
```

# Penjelasan Kode: Cropping Gambar Grayscale

Kode berikut digunakan untuk **memotong (crop)** bagian tertentu dari gambar grayscale, lalu menampilkannya dengan **Matplotlib**.
crop = grayscale[600:603, 600:603]
→ Melakukan cropping pada citra grayscale.
→ Format grayscale[y1:y2, x1:x2] digunakan untuk memilih bagian tertentu dari gambar:

600:603 pada sumbu y (baris).

600:603 pada sumbu x (kolom).
→ Artinya, kita hanya mengambil area kecil dari koordinat (600,600) hingga (602,602).
→ Hasil potongan disimpan ke dalam variabel crop.

plt.imshow(crop, 'gray')
→ Menampilkan hasil cropping (crop) dalam mode grayscale menggunakan Matplotlib.
→ Argumen 'gray' memastikan gambar ditampilkan dengan skema warna hitam putih.

```python
crop = grayscale[600:603,600:603]
plt.imshow(crop,'gray')
```

# Penjelasan Kode: Konversi Gambar Grayscale ke Format Unsigned Integer (uint8)

Kode berikut digunakan untuk **mengubah gambar grayscale** yang semula dalam format float (nilai 0–1) menjadi **format uint8 (0–255)** yang umum digunakan untuk representasi gambar digital.
import numpy as np
→ Mengimpor library NumPy dengan alias np.
→ NumPy digunakan untuk operasi numerik dan manipulasi array pada Python.

img_uint8 = (grayscale * 255).astype(np.uint8)
→ grayscale sebelumnya berbentuk array dengan nilai intensitas piksel antara 0.0 dan 1.0 (tipe float).
→ grayscale * 255 mengubah rentang nilai dari 0–1 menjadi 0–255.
→ .astype(np.uint8) mengubah tipe data array menjadi unsigned 8-bit integer (uint8), yang merupakan format standar penyimpanan gambar digital.
→ Hasil konversi disimpan dalam variabel img_uint8.

print(img_uint8)
→ Menampilkan array img_uint8 ke terminal.
→ Nilainya berupa angka-angka dengan rentang 0 sampai 255, yang mewakili intensitas grayscale setiap piksel.

# Penjelasan Kode: Konversi Gambar Grayscale ke Format uint8

Kode ini digunakan untuk **mengubah gambar grayscale** dari format float (rentang nilai 0–1) menjadi **format uint8 (rentang nilai 0–255)**, lalu menampilkannya dalam bentuk array.
import numpy as np
→ Mengimpor library NumPy dengan alias np.
→ NumPy dipakai untuk manipulasi array dan operasi numerik.

img_uint8 = (grayscale * 255).astype(np.uint8)
→ Variabel grayscale adalah gambar dalam format float dengan nilai piksel antara 0.0 hingga 1.0.
→ grayscale * 255 mengubah rentang nilai dari 0–1 menjadi 0–255 (sesuai standar nilai intensitas piksel).
→ .astype(np.uint8) mengubah tipe data menjadi unsigned 8-bit integer (uint8), yaitu tipe data umum untuk menyimpan gambar digital.
→ Hasil konversi disimpan dalam variabel img_uint8.

print(img_uint8)
→ Menampilkan isi array img_uint8 ke terminal.
→ Nilainya berupa matriks angka dengan rentang 0 sampai 255, yang mewakili intensitas grayscale setiap piksel.

```python
import numpy as np
img_uint8 = (grayscale * 255).astype(np.uint8)
print(img_uint8)
```

# Penjelasan Kode: Menampilkan Dua Gambar dalam Satu Figure

Kode berikut digunakan untuk membuat **satu figure dengan 1 baris dan 2 kolom** (dua subplot), lalu menampilkan dua gambar berbeda secara berdampingan menggunakan **Matplotlib**.
fig, axes = plt.subplots(1, 2, figsize=(10, 5))
→ Membuat sebuah figure dengan 1 baris dan 2 kolom (total 2 subplot).
→ figsize=(10, 5) mengatur ukuran figure menjadi lebar 10 inch dan tinggi 5 inch.
→ Variabel axes berupa array berisi dua objek subplot (akses lewat axes[0] dan axes[1]).

axes[0].imshow(crop2, cmap='gray')
→ Menampilkan gambar pertama (crop2) pada subplot pertama.
→ cmap='gray' memastikan gambar ditampilkan dalam skala grayscale.

axes[0].set_title("Gambar 1")
→ Memberikan judul "Gambar 1" pada subplot pertama.

axes[1].imshow(imageplus, cmap='gray')
→ Menampilkan gambar kedua (imageplus) pada subplot kedua dalam skala grayscale.

axes[1].set_title("Gambar 2")
→ Memberikan judul "Gambar 2" pada subplot kedua.

plt.tight_layout()
→ Mengatur tata letak subplot agar tidak saling menimpa atau terlalu rapat.

plt.show()
→ Menampilkan seluruh figure beserta kedua subplot ke layar.

```python
# Buat figure dengan 1 baris 2 kolom
fig, axes = plt.subplots(1, 2, figsize=(10, 5))

# Tampilkan gambar pertama
axes[0].imshow(crop2, cmap='gray')
axes[0].set_title("Gambar 1")

# Tampilkan gambar kedua
axes[1].imshow(imageplus, cmap='gray')
axes[1].set_title("Gambar 2")

plt.tight_layout()
plt.show()
```

# Splitting channels
Saat menggunakan perangkat lunak lain seperti ImageJ, langkah pertama yang sangat umum dalam jalur pemrosesan adalah membagi gambar menjadi saluran terpisah sehingga dapat diproses secara individual. Tidak ada "fungsi pemisahan" di sini, tetapi karena kita berurusan dengan array, kita dapat dengan mudah memotong array dan memilih hanya salah satu saluran. Jika kita ingin saluran pertama, kita perlu:semua baris dari dimensi pertama:semua kolom dari dimensi kedua:saluran pertama dari dimensi ketiga: 0

- Pengertiannya, keggunaannya

```python
image.shape
```

chanelRed = image[:, :, 0]
→ Variabel image adalah gambar berwarna dalam format RGB (Red, Green, Blue).
→ Format array image[:, :, channel]:

: pertama → mengambil semua baris (sumbu Y).

: kedua → mengambil semua kolom (sumbu X).

0 → mengambil channel ke-0, yaitu channel merah (Red).
→ Hasil ekstraksi disimpan dalam variabel chanelRed.

plt.imshow(chanelRed, 'gray')
→ Menampilkan hasil ekstraksi channel merah.
→ Parameter 'gray' digunakan agar nilai intensitas channel merah ditampilkan dalam skala abu-abu, bukan berwarna.

```python
chanelRed = image[:,:,0]
plt.imshow(chanelRed,'gray')
```

chanelGreen = image[:, :, 1]
→ Variabel image adalah gambar berwarna dalam format RGB (Red, Green, Blue).
→ Format array image[:, :, channel]:

: pertama → mengambil semua baris (sumbu Y).

: kedua → mengambil semua kolom (sumbu X).

1 → mengambil channel ke-1, yaitu channel hijau (Green).
→ Hasil ekstraksi disimpan dalam variabel chanelGreen.

plt.imshow(chanelGreen, 'gray')
→ Menampilkan hasil ekstraksi channel hijau.
→ Parameter 'gray' digunakan agar nilai intensitas channel hijau ditampilkan dalam skala abu-abu, bukan berwarna.

Kesimpulan

Kode ini berfungsi untuk mengambil channel hijau dari gambar RGB dan menampilkannya dalam mode grayscale.
Teknik ini berguna untuk analisis distribusi warna hijau atau pemrosesan citra berbasis channel tertentu.


Mau aku langsung bikinin sekalian dokumentasi untuk **channel biru (Blue)** juga biar lengkap semua channel RGB?


```python
chanelGreen = image[:,:,1]
plt.imshow(chanelGreen,'gray')
```

```python
chanelBlue = image[:,:,2]
plt.imshow(chanelBlue,'gray')
```

rgb_image = np.dstack((chanelRed, chanelGreen, chanelBlue))
→ Fungsi np.dstack() digunakan untuk menumpuk (stacking) tiga array 2D (chanelRed, chanelGreen, chanelBlue) di sepanjang sumbu kedalaman.
→ Hasilnya adalah gambar berwarna dengan format RGB.
→ Jika channel berasal dari gambar asli, nilainya bisa berupa float dengan rentang [0,1].

# rgb_image_uint8 = (rgb_image * 255).astype(np.uint8)
→ Baris ini dikomentari, tapi menunjukkan cara mengonversi jika array masih dalam format float.
→ rgb_image * 255 → mengubah nilai dari [0,1] menjadi [0,255].
→ .astype(np.uint8) → mengubah tipe data ke uint8, yaitu format standar penyimpanan gambar digital.
→ Perlu dilakukan jika ingin menyimpan hasil rekonstruksi ke file (misalnya .jpg atau .png).

plt.imshow(rgb_image)
→ Menampilkan gambar hasil rekonstruksi RGB menggunakan Matplotlib.
→ Jika data masih float dalam rentang [0,1], Matplotlib dapat langsung menampilkannya.

```python
rgb_image = np.dstack((chanelRed, chanelGreen, chanelBlue))  # hasil float [0,1]
# rgb_image_uint8 = (rgb_image * 255).astype(np.uint8) # Jika nilainya float maka harus konversi
plt.imshow(rgb_image)
```

fig, axes = plt.subplots(2, 2, figsize=(16, 14))
→ Membuat figure dengan ukuran 16×14 inch yang terdiri dari 2 baris × 2 kolom subplot.
→ Variabel axes berupa array 2D untuk mengakses subplot (axes[baris, kolom]).

Channel Red

axes[0,0].imshow(chanelRed, cmap='gray')
axes[0,0].set_title("Gambar Channel Red")
axes[0,0].axis("off")


→ Menampilkan channel merah dalam mode grayscale.
→ Judul ditambahkan, dan axis("off") menyembunyikan sumbu koordinat.

Channel Green

axes[0,1].imshow(chanelGreen, cmap='gray')
axes[0,1].set_title("Gambar Channel Green")
axes[0,1].axis("off")


→ Menampilkan channel hijau dalam mode grayscale.

Channel Blue

axes[1,0].imshow(chanelBlue, cmap='gray')
axes[1,0].set_title("Gambar Channel Blue")
axes[1,0].axis("off")


→ Menampilkan channel biru dalam mode grayscale.

Rekonstruksi RGB

axes[1,1].imshow(rgb_image)
axes[1,1].set_title("Gambar Rekonstruksi RGB")
axes[1,1].axis("off")


→ Menampilkan hasil penggabungan channel Red, Green, dan Blue kembali menjadi gambar RGB.
→ Tidak menggunakan cmap='gray' karena ini gambar berwarna.

plt.tight_layout()
→ Mengatur tata letak agar subplot tidak saling bertumpuk.

plt.show()
→ Menampilkan semua subplot dalam satu figure.

```python
# Buat figure dengan 1 baris 2 kolom
fig, axes = plt.subplots(2, 2, figsize=(16, 14))

# Tampilkan gambar pertama
axes[0,0].imshow(chanelRed, cmap='gray')
axes[0,0].set_title("Gambar Channel Red")
axes[0,0].axis("off")

# Tampilkan gambar kedua
axes[0,1].imshow(chanelGreen, cmap='gray')
axes[0,1].set_title("Gambar Channel reen")
axes[0,1].axis("off")

# Tampilkan gambar pertama
axes[1,0].imshow(chanelBlue, cmap='gray')
axes[1,0].set_title("Gambar Channel Blue")
axes[1,0].axis("off")

# Tampilkan gambar kedua
axes[1,1].imshow(rgb_image, cmap='gray')
axes[1,1].set_title("Gambar 2")
axes[1,1].axis("off")
plt.tight_layout()
plt.show()
```
